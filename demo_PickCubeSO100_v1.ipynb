```python
# Must be FIRST — before any mani_skill import
import os
os.environ["MANI_SKILL_DATA_DIR"] = "/root/.maniskill/data"

# Import required packages
import gymnasium as gym
import torch
import mani_skill.envs
from tqdm.notebook import tqdm
from mani_skill.utils.wrappers import RecordEpisode

env = gym.make("PickCubeSO100-v1", num_envs=8, render_mode="rgb_array", enable_shadow=True, reward_mode='dense')
env = RecordEpisode(
    env,
    "./videos",
    max_steps_per_video=100
)

obs, _ = env.reset()
for i in tqdm(range(100)):
    action = env.action_space.sample()
    obs, reward, terminated, truncated, info = env.step(torch.from_numpy(action))
    # print(reward)
env.close()

from IPython.display import Video
Video("./videos/0.mp4", embed=True, width=640)
```
